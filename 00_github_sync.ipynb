```python
import os
from google.colab import userdata

# Retrieve secrets securely from Colab Secrets Manager
token = userdata.get('GITHUB_TOKEN')
username = userdata.get('GITHUB_USERNAME')
email = userdata.get('GITHUB_EMAIL')
repo_name = "macro_risk_pipeline"

# Configure Git Global Identity
!git config --global user.email "{email}"
!git config --global user.name "{username}"

# Construct the secure clone URL
repo_url = f"https://{username}:{token}@github.com/{username}/{repo_name}.git"

# Clone or pull the repository
if not os.path.exists(repo_name):
    !git clone {repo_url}
    print(f"Successfully cloned {repo_name}")
else:
    print(f"{repo_name} already exists locally. Pulling latest updates...")
    %cd {repo_name}
    !git pull
    %cd ..

# Change working directory into the repository
os.chdir(repo_name)
print(f"Current working directory: {os.getcwd()}")
```
